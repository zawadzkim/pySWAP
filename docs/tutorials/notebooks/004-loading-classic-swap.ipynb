# Load classic SWAP model

This notebook loads an existing SWAP model from its ASCII input files into pySWAP classes.

`Model.load()` reads the .swp file and the files it references in the same directory: the crop files (.crp) in the crop rotation, the drainage file (.dra), the bottom boundary file (.bbc), the irrigation file (.irg) and the weather data (a single .met file, or yearly files). We use the grass growth testcase that ships with SWAP, in which the bottom boundary is defined in a separate .bbc file.

```python
import pyswap as psp
```

```python
meta = psp.components.Metadata(
    author="John Doe",
    institution="University of Somewhere",
    email="john.doe@somewhere.com",
    project="pySWAP test - grassgrowth",
    swap_ver="4.2",
)

ml = psp.Model.load(psp.testcase.get_path("grassgrowth", "swp"), metadata=meta)
```

The model now holds every component, including those that came from the other files: the bottom boundary from the .bbc file, the drainage settings from the .dra file and the grass crop file.

```python
print(ml.swp)
```

```python
print(ml.crop.cropfiles["grassd"].crp)
```

```python
ml.meteorology.metfile.content.head()
```

```python
result = ml.run(silence_warnings=True)
result.yearly_summary
```

## Loading single files

Each file can also be loaded on its own, e.g. to replace one component of a model. The loaders are `load_swp`, `load_crp`, `load_dra`, `load_bbc`, `load_irg` and `load_met`. `load_met` reads a SWAP weather file into a `MetFile`; given a list of yearly files (`<name>.yyy`), it merges them into one table.

```python
metfile = psp.load_met(psp.testcase.get_path("grassgrowth", "met"), metfil="260.met")
ml.meteorology.metfile = metfile
metfile.content.tail()
```
